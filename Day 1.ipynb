## Set Matrix Zeroes

**Problem sources**  
- [LeetCode – Set Matrix Zeroes](https://leetcode.com/problems/set-matrix-zeroes/)

### Problem Statement (In Your Own Words)
You are given an $m \times n$ integer matrix `matrix`.

- **If an element is 0, you must set its entire row and column to 0.**  
- This transformation should be done **in-place** (i.e., modify the original matrix, don’t return a new one).

Your goal is to transform the matrix so that all rows and columns that originally contained at least one `0` become completely zeroed in the final matrix.

---

### Quick Example
**Input**  
`matrix = [[1,1,1], [1,0,1], [1,1,1]]`

**Output**  
`[[1,0,1], [0,0,0], [1,0,1]]`

**Why?**  
- The `0` at position `(1,1)` (0-based indices) makes **row 1** and **column 1** all zeros.

---

### Constraints (What You Must Respect)
- **Matrix size**: $1 \le m, n \le 200$
- **Values**: `-2^31 <= matrix[i][j] <= 2^31 - 1`
- **Must** modify the matrix in-place.
- Try to achieve **$O(1)$ extra space** for the optimal solution.

### Approach 1 – Using Extra Space (Easier to Think About)
**Idea**: Use two helper arrays to remember which rows and columns should become zero.

- **Step 1**: Create `row[m]` and `col[n]`, initially all `0`.
- **Step 2**: First pass through the matrix:
  - If `matrix[i][j] == 0`, mark `row[i] = 1` and `col[j] = 1`.
- **Step 3**: Second pass:
  - If `row[i] == 1` **or** `col[j] == 1`, set `matrix[i][j] = 0`.

**Complexity**:  
- **Time**: $O(m \times n)$  
- **Space**: $O(m + n)$ extra

### Approach 2 – Optimal In-Place (Using First Row & First Column as Markers)
This is the main idea emphasized in [takeuforward – Set Matrix Zero](https://takeuforward.org/data-structure/set-matrix-zero).

**Key trick**:  
Use the **first row and first column** as storage for which rows/columns should be zeroed, instead of separate arrays.

#### High-Level Steps
1. **Detect if first row and first column need to be zeroed at the end**:
   - Use two booleans, e.g. `firstRowZero`, `firstColZero`.
2. **Mark the rest of the matrix** (excluding first row/col):
   - Whenever you see `matrix[i][j] == 0` for `i > 0` and `j > 0`:
     - Set `matrix[i][0] = 0` (mark row `i` to be zeroed).
     - Set `matrix[0][j] = 0` (mark column `j` to be zeroed).
3. **Zero out based on markers**:
   - For all cells `i > 0`, `j > 0`:
     - If `matrix[i][0] == 0` **or** `matrix[0][j] == 0`, set `matrix[i][j] = 0`.
4. **Finally handle first row and first column**:
   - If `firstRowZero` is true → set entire first row to 0.
   - If `firstColZero` is true → set entire first column to 0.

**Complexity**:  
- **Time**: $O(m \times n)$  
- **Extra space**: $O(1)$ (ignoring the input matrix)

---

### Pseudo-Code (For the Optimal In-Place Approach)
Try to **fill in the details yourself first**, then compare with a reference.

```text
1. Set firstRowZero = false, firstColZero = false

2. Check first row:
   - If any element in row 0 is 0, set firstRowZero = true

3. Check first column:
   - If any element in column 0 is 0, set firstColZero = true

4. Use first row & col as markers:
   - For i from 1 to m-1:
       For j from 1 to n-1:
           If matrix[i][j] == 0:
               matrix[i][0] = 0
               matrix[0][j] = 0

5. Zero cells based on markers:
   - For i from 1 to m-1:
       For j from 1 to n-1:
           If matrix[i][0] == 0 OR matrix[0][j] == 0:
               matrix[i][j] = 0

6. Handle first row:
   - If firstRowZero == true:
       For j from 0 to n-1:
           matrix[0][j] = 0

7. Handle first column:
   - If firstColZero == true:
       For i from 0 to m-1:
           matrix[i][0] = 0
```

## MY Solution

````python
class Solution(object):
    def setZeroes(self, matrix):
        """
        :type matrix: List[List[int]]
        :rtype: void Do not return anything, modify matrix in-place instead.
        """
        R = len(matrix)
        C = len(matrix[0])
        rows, cols = set(), set()

        # Essentially, we mark the rows and columns that are to be made zero
        for i in range(R):
            for j in range(C):
                if matrix[i][j] == 0:
                    rows.add(i)
                    cols.add(j)

        # Iterate over the array once again and using the rows and cols sets, update the elements
        for i in range(R):
            for j in range(C):
                if i in rows or j in cols:
                    matrix[i][j] = 0
````

## Pascal's Triangle

**Problem sources**  
- [LeetCode – Pascal's Triangle](https://leetcode.com/problems/pascals-triangle/)

### Problem Statement
Given an integer `numRows`, return the first `numRows` of Pascal's triangle. Each row of the triangle starts and ends with `1`. Every other element is the sum of the two elements directly above it from the previous row.

### Example
**Input**: `numRows = 5`

**Output**: `[[1], [1,1], [1,2,1], [1,3,3,1], [1,4,6,4,1]]`

---

### Constraints
- `1 <= numRows <= 30`

---

### Approaches

- **Iterative (build row-by-row)**: Start with the first row `[1]`. For each subsequent row k, create a new row of length k where the first and last elements are `1`. For indices 1..k-2 set `row[j] = prev[j-1] + prev[j]` using the previous row. This is simple, efficient, and easy to implement.

- **Combinatorics (direct formula)**: The j-th element (0-based) of row i is `C(i, j)` (binomial coefficient). Use iterative multiplicative formula to compute each entry in O(1) amortized per entry without full factorials: `val = val * (i - j + 1) // j` pattern when generating in a row. This avoids using the previous row but is slightly more math-heavy.

---

### Pseudocode (Iterative)
```text
if numRows == 0: return []
triangle = [[1]]
for i from 1 to numRows-1:
    prev = triangle[-1]
    row = [1]
    for j from 1 to i-1:
        row.append(prev[j-1] + prev[j])
    row.append(1)
    triangle.append(row)
return triangle
```

### Complexity
- Time: `O(numRows^2)` — you must produce about numRows*(numRows+1)/2 elements.
- Space: `O(numRows^2)` for the output; iterative generation can use `O(numRows)` extra if streaming rows instead of storing all.

---

### Notes & Tips
- For LeetCode-style problems return the full triangle as a list of lists. For streaming or memory-constrained contexts, produce rows one-by-one and discard previous rows when not needed.
- For large `numRows`, prefer the combinatorial formula if you need random access to specific entries without constructing the entire triangle.

---

**Quick sample dry-run**: `numRows = 3` → start `[1]` → next `[1,1]` → next `[1,2,1]` → output `[[1],[1,1],[1,2,1]]`.

````python
#Solution 1
class Solution(object):
    def generate(self, numRows):
        """
        :type numRows: int
        :rtype: List[List[int]]
        """
        if numRows == 0:
            return []
        elif numRows == 1:
            return [[1]]
        elif numRows == 2:
            return [[1], [1, 1]]

        triangle = [[1], [1, 1]]

        for i in range(1, numRows-1):
            curr = triangle[i]
            list = [1]
            for j in range(len(curr) - 1):
                k = curr[j] + curr[j + 1]
                list.append(k)
            list.append(1)
            triangle.append(list)

        return triangle



#Solution 2

def fact(n):
    if n<= 1:
        return 1
    factorial = 1
    for i in range(1, n+1):
        factorial *= i
    return int(factorial)

def combination(n, m): #4,2 
    return int((fact(n))/(fact(m)*fact(n-m)))


class Solution(object):
    def generate(self, numRows):
        pascalTriangle = []
        for i in range(numRows):
            arr = []
            for j in range(i+1):
                arr.append(combination(i, j))
            pascalTriangle.append(arr)
        return pascalTriangle
        
    
````
